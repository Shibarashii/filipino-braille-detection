Install `requirements.txt`


```python
import subprocess
import pkg_resources

with open("requirements.txt") as f:
    for line in f:
        pkg = line.strip()
        if not pkg or pkg.startswith("#"):
            continue
        pkg_name = pkg.split("==")[0]
        try:
            pkg_resources.get_distribution(pkg_name)
        except pkg_resources.DistributionNotFound:
            print(f"Installing missing package: {pkg}")
            subprocess.check_call(["pip", "install", pkg])
```

## Download Outputs


```python
import shutil
from pathlib import Path
from google.colab import files
import zipfile
from datetime import datetime


def download_all_outputs(experiment_name=None):
    """
    Download all training outputs as a single zip file

    Args:
        experiment_name: Specific experiment folder name, or None for all
    """
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    if experiment_name:
        source_dir = Path(f"outputs/models/{experiment_name}")
        zip_name = f"{experiment_name}_{timestamp}.zip"
    else:
        source_dir = Path("outputs")
        zip_name = f"all_outputs_{timestamp}.zip"

    if not source_dir.exists():
        print(f"❌ Directory not found: {source_dir}")
        return

    print(f"📦 Creating zip file: {zip_name}")
    print(f"📁 Source: {source_dir}")

    shutil.make_archive(
        zip_name.replace('.zip', ''),
        'zip',
        source_dir.parent,
        source_dir.name
    )

    print(f"⬇️  Downloading {zip_name}...")
    files.download(zip_name)
    print("✅ Download complete!")


download_all_outputs()
```

```python
import shutil
from pathlib import Path
from google.colab import files
import zipfile
from datetime import datetime


def download_runs_directory():
    """
    Download all contents from the runs/ directory as a single zip file
    """
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    source_dir = Path("runs")
    zip_name = f"runs_backup_{timestamp}.zip"

    if not source_dir.exists():
        print(f"❌ Directory not found: {source_dir}")
        return

    # Check if directory has content
    content = list(source_dir.rglob('*'))
    if not content:
        print(f"⚠️  Directory is empty: {source_dir}")
        return

    print(f"📦 Creating zip file: {zip_name}")
    print(f"📁 Source: {source_dir}")
    print(f"📊 Found {len(content)} items to archive")

    # Create the zip file
    shutil.make_archive(
        zip_name.replace('.zip', ''),
        'zip',
        source_dir.parent,
        source_dir.name
    )

    # Get file size
    zip_path = Path(zip_name)
    size_mb = zip_path.stat().st_size / (1024 * 1024)
    print(f"📦 Zip file size: {size_mb:.2f} MB")

    print(f"⬇️  Downloading {zip_name}...")
    files.download(zip_name)
    print("✅ Download complete!")
    print(f"\n📂 Your runs/ directory backup is ready!")


# Execute the download
download_runs_directory()
```
